```python
import numpy as np
import scipy.constants as cst
```

We look at an ensemble of atoms trapped in a tweezer trap. We know the approximate temperature of the atoms and trapping frequency, so we aim to estimate the average phonon number.

It is known that the energy of a motional state with a definite number of $n$ phonons is:

$$E_n = \hbar \omega_t (n + 1/2)$$

But, assuming the atomic ensemble is in a thermal state, then is it not in a pure motional eigenstate, rather it is an incoherent superposition of different eigenstates.

$$
\rho_{th} = \sum_n \exp(-E_n/k_B T) |n\rangle \langle n|
$$

It is known the average phonon number for such a thermal state is:

$$
\bar n = \frac{1}{\exp(\hbar \omega_t /k_B T) - 1}
$$

```python
T = 10e-6 # 10 uK temperature
w_t = 100e3 # 100 kHz trapping frequency

n_bar = 1 / (np.exp(cst.hbar * w_t / (cst.k * T)) - 1)

print(f"Average phonon number: {n_bar:.3f} phonons")

print(f"Low temperature approximation: {cst.k * T / (cst.hbar * w_t):.3f} phonons")
```
