```python
import numpy as np
from arc import *
```

```python
atom = Ytterbium174()
```

```python
n = 6
l = 0 # P
j = 0 # J
calc = DynamicPolarizability(atom, n, 1, j)

# minimal and maximal n for the basis set over which we count contributions
calc.defineBasis(atom.groundStateN, n+15)
print(atom.groundStateN)

wavelengthList = np.linspace(450, 800, 1000) * 1e-9

ax = calc.plotPolarizability(
    wavelengthList, debugOutput=True, units="au", line="r--"
)
```
